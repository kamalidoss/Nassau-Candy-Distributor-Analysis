# **Step 1: Data Cleaning**

```python
#Convert Dates

df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')
```

```python
#Calculate Lead Time

df['Lead_Time'] = (
    df['Ship Date'] - df['Order Date']
).dt.days
```

```python
#Remove Invalid Records

df = df[df['Lead_Time'] >= 0]
```

```python
df.info()
```

# **Step 2: Factory Mapping**

The dataset doesn't directly contain factory names.

Factory must be derived using Product Name.

```python
factory_map = {
    "Wonka Bar - Nutty Crunch Surprise":"Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows":"Lot's O' Nuts",
    "Wonka Bar -Scrumdiddlyumptious":"Lot's O' Nuts",

    "Wonka Bar - Milk Chocolate":"Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel":"Wicked Choccy's",

    "Laffy Taffy":"Sugar Shack",
    "SweeTARTS":"Sugar Shack",
    "Nerds":"Sugar Shack",
    "Fun Dip":"Sugar Shack",
    "Fizzy Lifting Drinks":"Sugar Shack"
}
```

```python
df["Factory"] = df["Product Name"].map(factory_map)
```

# **Step 3: Route Definition**

```python
#Route by State

df["Route_State"] = (
    df["Factory"] + " → " +
    df["State/Province"]
)
```

```python
#Route by Region

df["Route_Region"] = (
    df["Factory"] + " → " +
    df["Region"]
)
```

**Route Efficiency Analysis**

```python
#Route Performance Table

route_perf = (
    df.groupby("Route_State")
    .agg({
        "Lead_Time":["mean","std","count"]
    })
)
```

```python
#Top 10 Efficient Routes

route_perf.sort_values(
    ('Lead_Time','mean')
).head(10)
```

```python
#Bottom 10 Inefficient Routes

route_perf.sort_values(
    ('Lead_Time','mean'),
    ascending=False
).head(10)
```

**Geographic Bottleneck Analysis**

```python
#State-Level Delays

state_delay = (
    df.groupby("State/Province")
    ["Lead_Time"]
    .mean()
)
```

```python
#Region Performance

region_perf = (
    df.groupby("Region")
    ["Lead_Time"]
    .mean()
)
```

**Ship Mode Analysis**

```python
#Average Lead Time by Ship Mode

df.groupby("Ship Mode")["Lead_Time"].mean()
```

```python
#Ship Mode vs Profit

df.groupby("Ship Mode").agg({
    "Sales":"sum",
    "Gross Profit":"sum",
    "Lead_Time":"mean"
})
```

**Route Efficiency Score**

```python
#Normalize Lead Time

max_lead_time = df['Lead_Time'].max()
df['Efficiency_Score'] = 100 - (df['Lead_Time'] / max_lead_time) * 100
```

```python
max_time = df["Lead_Time"].max()

df["Efficiency_Score"] = (
100 -
(df["Lead_Time"]/max_time)*100
)
```
